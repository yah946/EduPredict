```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split,GridSearchCV,KFold
from sklearn.metrics import mean_squared_error,r2_score

df = pd.read_csv('cleaned_dataset.csv')
```

```python
X = df.drop('Exam_Score',axis=1)
y = df['Exam_Score']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
my_model = RandomForestRegressor()
my_model.fit(X_train,y_train)
```

```python
y_pred = my_model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = mse**.5
r2 = my_model.score(X_test,y_test)
print(f'The mse value is: {mse},\nand thier of rmse is: {rmse},\nmy model score: {round(r2*100,2)}%')
```

```python
cv = KFold(n_splits = 3,shuffle=True,random_state=10)
CV_rfr = GridSearchCV(estimator=my_model, param_grid={'n_estimators': [300,600,900],'min_samples_split': [20,100,500],'max_depth' : [3,5,7]}, cv= cv,scoring='r2')
CV_rfr.fit(X_train, y_train)
```

```python
CV_rfr.cv_results_
```

```python
CV_rfr.best_params_
```

```python
best_model = CV_rfr.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
```

```python
r2 = r2_score(y_test,y_pred)
print(f'r2 score is: {round(r2*100,2)}%')
```
